```python
import sys
class Employee:
    def show(self):
        print(f"The python version is {sys.version}")
    def __init__ (self,name:str,address:str,phone:int): #constructor
        print("-----Constructor-----")
        self.name = name
        self.address = address
        self.phone = phone
    def __str__(self):
        return f"Name is {self.name} Address is {self.address} and phone is {self.phone}."


e1 = Employee('My','Name',1233)
print(e1.__str__())
print(e1)
```
